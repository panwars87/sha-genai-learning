# Module 00: Course setup (Colab)

This notebook replaces the Mac steps. Instead of a `.env` file, you store secrets in **Colab Secrets**: click the **key icon** in the left sidebar, add each name below, and switch on "Notebook access".

| Secret name | What it is |
|---|---|
| `ANTHROPIC_API_KEY` | From console.anthropic.com |
| `OPENAI_API_KEY` | From platform.openai.com |
| `LITELLM_MASTER_KEY` | A long random string you invent, starting with `sk-` |
| `DATABASE_URL` | Your free hosted Postgres URL (e.g. Neon) |
| `GITHUB_TOKEN` | Only if the course repo is private |

Secret values are never printed by this notebook.

```python
# ── Get the course code ──────────────────────────────────────────────────────
# BRANCH: "main" once the course is merged; otherwise the branch you are studying.
REPO = "github.com/panwars87/sha-genai-learning.git"
BRANCH = "main"

import os, subprocess
from google.colab import userdata

def secret(name, required=True):
    """Read a Colab Secret by name (never printed)."""
    try:
        return userdata.get(name)
    except Exception:
        if required:
            raise SystemExit(f"Add '{name}' in Colab Secrets (key icon, left sidebar) and allow notebook access.")
        return None

# A private repo needs a GitHub token secret named GITHUB_TOKEN (read-only is enough).
token = secret("GITHUB_TOKEN", required=False)
url = f"https://{token}@{REPO}" if token else f"https://{REPO}"
if not os.path.isdir("/content/sha-genai-learning"):
    subprocess.run(["git", "clone", "-q", "-b", BRANCH, url, "/content/sha-genai-learning"], check=True)
%cd /content/sha-genai-learning
!pip -q install uv
print("Ready.")
```

## L02 · Secrets (the Colab version of the `.env` check)
Checks that each secret exists. It shows only names, never values.

```python
import tomllib
names = tomllib.load(open("modules/00-course-setup/lessons/checks.toml", "rb"))["env_file"]
for name in names["required"]:
    value = secret(name, required=False)
    print(("  ✓ " if value else "  ✗ ") + f"{name:<22}" + ("set" if value else "MISSING"))
mk = secret("LITELLM_MASTER_KEY", required=False) or ""
print(("  ✓ " if mk.startswith(names["master_key_prefix"]) else "  ✗ ") + "master key starts with 'sk-'")
```
